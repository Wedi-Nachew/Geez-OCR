## 3. Train the Models using K-fold
I am going to use `Early Stopping` to stop training automatically as soon as validation loss stops improving.

```python
early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=15,
    restore_best_weights=True,
    verbose=1
)
```

### 3.1. Train ANN Models using K-folds

#### 3.1.1. Simpler ANN Model

```python
ann_results = {}

print("=" * 50)
print("Simpler ANN Model")
print("=" * 50)
simpler_ann_results = cross_validate_model(
    build_simpler_ann,
    X_train,
    y_train,
    epochs=300,
    callback=early_stopping
)
```

```python
# Mean of the results of the 5 fold trainings of the model
summarized_results = summarize_results(simpler_ann_results)
print(f"Mean Train Accuracy: {summarized_results["mean_train_accuracy"]}")
print(f"Mean Cross-Validation Accuracy: {summarized_results["mean_cv_accuracy"]}")
print(f"STD of Training Accuracy: {summarized_results["std_train_accuracy"]}")
print(f"STD of Cross-Validation Accuracy: {summarized_results["std_cv_accuracy"]}")
print(f"Mean Gap: {summarized_results["mean_gap"]}")
print(f"Mean Train Top 5 Accuracy: {summarized_results["mean_train_top5_accuracy"]}")
print(f"Mean Cross-Validation Top 5 Accuracy: {summarized_results["mean_cv_top5_accuracy"]}")

# Add the results to the ann results
ann_results["Simpler ANN Model"] = summarized_results
```

#### 3.1.2. Deeper ANN Model

```python
print("=" * 50)
print("Deeper ANN Model")
print("=" * 50)
deeper_ann_results = cross_validate_model(
    build_deeper_ann,
    X_train,
    y_train,
    epochs=300,
    callback=early_stopping
)
```

```python
# Mean of the results of the 5 fold trainings of the model
summarized_results = summarize_results(deeper_ann_results)
print(f"Mean Train Accuracy: {summarized_results["mean_train_accuracy"]}")
print(f"Mean Cross-Validation Accuracy: {summarized_results["mean_cv_accuracy"]}")
print(f"STD of Training Accuracy: {summarized_results["std_train_accuracy"]}")
print(f"STD of Cross-Validation Accuracy: {summarized_results["std_cv_accuracy"]}")
print(f"Mean Gap: {summarized_results["mean_gap"]}")
print(f"Mean Train Top 5 Accuracy: {summarized_results["mean_train_top5_accuracy"]}")
print(f"Mean Cross-Validation Top 5 Accuracy: {summarized_results["mean_cv_top5_accuracy"]}")

# Add the results to the ann results
ann_results["Deeper ANN Model"] = summarized_results
```

#### 3.1.3. Regularized Deeper ANN Model

```python
print("=" * 50)
print("Regularized Deeper ANN Model")
print("=" * 50)
regularized_deeper_ann_results = cross_validate_model(
    build_regularized_deeper_ann,
    X_train,
    y_train,
    epochs=300,
    callback=early_stopping
)
```

```python
# Mean of the results of the 5 fold trainings of the model
summarized_results = summarize_results(regularized_deeper_ann_results)
print(f"Mean Train Accuracy: {summarized_results["mean_train_accuracy"]}")
print(f"Mean Cross-Validation Accuracy: {summarized_results["mean_cv_accuracy"]}")
print(f"STD of Training Accuracy: {summarized_results["std_train_accuracy"]}")
print(f"STD of Cross-Validation Accuracy: {summarized_results["std_cv_accuracy"]}")
print(f"Mean Gap: {summarized_results["mean_gap"]}")
print(f"Mean Train Top 5 Accuracy: {summarized_results["mean_train_top5_accuracy"]}")
print(f"Mean Cross-Validation Top 5 Accuracy: {summarized_results["mean_cv_top5_accuracy"]}")

# Add the results to the ann results
ann_results["Regularized Deeper ANN Model"] = summarized_results
```
